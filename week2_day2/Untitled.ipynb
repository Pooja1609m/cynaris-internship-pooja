```python
# First DataFrame: Sales data
sales_df = df.copy()

# Second DataFrame: Region information
region_df = pd.DataFrame({
    "region": sales_df["region"].dropna().unique(),
    "manager": ["Manager " + str(i + 1)
                for i in range(sales_df["region"].dropna().nunique())]
})

print("Sales Data:")
display(sales_df.head())

print("Region Data:")
display(region_df)
```

```python
inner_merge = pd.merge(
    sales_df,
    region_df,
    on="region",
    how="inner"
)

display(inner_merge.head())
```

```python
left_merge = pd.merge(
    sales_df,
    region_df,
    on="region",
    how="left"
)

display(left_merge.head())
```

```python
right_merge = pd.merge(
    sales_df,
    region_df,
    on="region",
    how="right"
)

display(right_merge.head())
```

```python
outer_merge = pd.merge(
    sales_df,
    region_df,
    on="region",
    how="outer"
)

display(outer_merge.head())
```

```python
import pandas as pd

pivot_data = pd.DataFrame({
    "region": ["North", "North", "South", "South", "East", "East"],
    "product": ["Laptop", "Mobile", "Laptop", "Mobile", "Laptop", "Mobile"],
    "sales_amount": [50000, 30000, 45000, 25000, 40000, 20000]
})

display(pivot_data)
```

```python
pivot_table = pd.pivot_table(
    pivot_data,
    values="sales_amount",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)

display(pivot_table)
```

```python
wide_data = pd.DataFrame({
    "Region": ["North", "South"],
    "January": [1000, 1200],
    "February": [1500, 1800],
    "March": [2000, 2200]
})

display(wide_data)
```

```python
long_data = pd.melt(
    wide_data,
    id_vars=["Region"],
    var_name="Month",
    value_name="Sales"
)

display(long_data)
```

```python
result = (
    pivot_data
    .dropna()
    .groupby("region")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

display(result)
```
