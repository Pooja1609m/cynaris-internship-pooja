```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the Seaborn chart style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../week1_day5/day5_sales.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
# Display column names
print("Column Names:")
print(df.columns)

# Check dataset information
print("\nDataset Information:")
df.info()

# Get basic statistics
print("\nDataset Statistics:")
display(df.describe())
```

```python
list(df.columns)
```

```python
print(df.head())
print(df.columns.tolist())
```

```python
df.select_dtypes(include="number").columns.tolist()
```

```python
plt.figure(figsize=(10, 5))

df.plot(kind="line", ax=plt.gca())

plt.title("Sales Data - Monthly Trend", fontsize=14)
plt.xlabel("Records")
plt.ylabel("Sales Values")

plt.annotate(
    "Sales Trend",
    xy=(0.5, 0.9),
    xycoords="axes fraction",
    fontsize=10,
    color="red"
)

plt.grid(True)
plt.tight_layout()
plt.savefig("line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df.select_dtypes(include="number").sum().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sales Data - Bar Chart")
plt.xlabel("Numeric Columns")
plt.ylabel("Total Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
numeric_columns = df.select_dtypes(include="number").columns

plt.figure(figsize=(8, 5))

plt.scatter(
    df[numeric_columns[0]],
    df[numeric_columns[1]],
    alpha=0.7
)

plt.title("Relationship Between Two Numeric Variables")
plt.xlabel(numeric_columns[0])
plt.ylabel(numeric_columns[1])

plt.tight_layout()
plt.savefig("scatter_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
numeric_columns = df.select_dtypes(include="number").columns

plt.figure(figsize=(8, 5))

plt.hist(df[numeric_columns[0]], bins=10, edgecolor="black")

plt.title("Distribution of Sales Data")
plt.xlabel(numeric_columns[0])
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("box_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
numeric_columns = df.select_dtypes(include="number").columns

plt.figure(figsize=(8, 5))

plt.boxplot(df[numeric_columns[0]].dropna())

plt.title("Sales Data - Box Plot")
plt.ylabel(numeric_columns[0])

plt.tight_layout()
plt.savefig("histogram.png", dpi=300, bbox_inches="tight")
plt.show()
```
