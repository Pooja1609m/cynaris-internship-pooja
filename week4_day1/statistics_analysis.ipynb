```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv(
    r"C:\Users\pooja\Downloads\cynaris\cynaris-internship-pooja\week2_day4\cleaned_dataset.csv"
)

# Display first five rows
df.head()
```

```python

plt.figure(figsize=(8, 5))
plt.hist(df["sales_amount"], bins=10, color="skyblue", edgecolor="black")
plt.title("Frequency Distribution of Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

# Compare sales distribution across regions

plt.figure(figsize=(9, 5))

df.boxplot(
    column="sales_amount",
    by="region",
    grid=False
)

plt.title("Sales Distribution Across Regions")
plt.suptitle("")
plt.xlabel("Region")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python

# Select the first two regions
regions = df["region"].dropna().unique()[:2]

# Filter sales data for those regions
comparison = df[df["region"].isin(regions)]

# Create side-by-side box plots
plt.figure(figsize=(8, 5))

comparison.boxplot(
    column="sales_amount",
    by="region",
    grid=False
)

plt.title("Sales Comparison Between Two Regions")
plt.suptitle("")
plt.xlabel("Region")
plt.ylabel("Sales Amount")
plt.show()

print("Regions compared:", regions)
```

```python

# Complete descriptive statistics summary
df["sales_amount"].describe()
```

```python

# Calculate skewness of sales
skewness = df["sales_amount"].skew()

print("Sales Skewness:", skewness)

if skewness > 0.5:
    print("Distribution: Right-skewed")
elif skewness < -0.5:
    print("Distribution: Left-skewed")
else:
    print("Distribution: Approximately symmetric")
```
