```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy version:", np.__version__)
print("NumPy imported successfully!")
```

```python
# Create a 1D array
sales = np.array([100, 200, 300, 400, 500])

# Display the array
print("Sales Array:", sales)

# Check dimensions
print("Dimensions:", sales.ndim)

# Check shape
print("Shape:", sales.shape)
```

```python
# Create a 2D array
sales_2d = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

# Display the array
print("2D Sales Array:")
print(sales_2d)

# Check dimensions
print("Dimensions:", sales_2d.ndim)

# Check shape
print("Shape:", sales_2d.shape)
```

```python
# Create a 3D array
sales_3d = np.array([
    [[100, 200], [300, 400]],
    [[500, 600], [700, 800]]
])

# Display the array
print("3D Sales Array:")
print(sales_3d)

# Check dimensions
print("Dimensions:", sales_3d.ndim)

# Check shape
print("Shape:", sales_3d.shape)
```

```python
# Original sales
sales = np.array([100, 200, 300, 400, 500])

# Add 50 to every value
print("Addition:", sales + 50)

# Multiply every value by 2
print("Multiplication:", sales * 2)

# Subtract 20 from every value
print("Subtraction:", sales - 20)

# Increase sales by 10%
print("Sales after 10% increase:", sales * 1.10)
```

```python
# Sales data
sales = np.array([100, 200, 300, 400, 500])

# Calculate total sales
print("Total Sales:", np.sum(sales))

# Calculate average sales
print("Average Sales:", np.mean(sales))

# Find maximum sales
print("Maximum Sales:", np.max(sales))

# Find minimum sales
print("Minimum Sales:", np.min(sales))

# Calculate standard deviation
print("Standard Deviation:", np.std(sales))
```

```python
# Sales of 3 products across 2 regions
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

# Calculate column-wise total
print("Column-wise Total:", np.sum(sales, axis=0))

# Calculate row-wise total
print("Row-wise Total:", np.sum(sales, axis=1))

# Calculate column-wise average
print("Column-wise Average:", np.mean(sales, axis=0))

# Calculate row-wise average
print("Row-wise Average:", np.mean(sales, axis=1))
```

```python
# Sales data
sales = np.array([100, 200, 300, 400, 500, 600])

# Calculate mean
print("Mean:", np.mean(sales))

# Calculate standard deviation
print("Standard Deviation:", np.std(sales))

# Calculate 50th percentile (median)
print("50th Percentile:", np.percentile(sales, 50))

# Calculate 90th percentile
print("90th Percentile:", np.percentile(sales, 90))
```

```python
# Advertising expenses (in thousands)
advertising = np.array([10, 20, 30, 40, 50])

# Sales (in thousands)
sales = np.array([100, 200, 300, 400, 500])

# Calculate correlation
correlation = np.corrcoef(advertising, sales)

print("Correlation Matrix:")
print(correlation)

print("Correlation Value:", correlation[0, 1])
```

```python
import time

# Create 1 million numbers
size = 1_000_000

# Python list
python_list = list(range(size))

# NumPy array
numpy_array = np.arange(size)

# Python loop performance
start = time.time()

python_result = []
for num in python_list:
    python_result.append(num * 2)

python_time = time.time() - start

# NumPy performance
start = time.time()

numpy_result = numpy_array * 2

numpy_time = time.time() - start

# Display results
print("Python Loop Time:", python_time, "seconds")
print("NumPy Time:", numpy_time, "seconds")

print("Python is slower by:",
      round(python_time / numpy_time, 2), "times")
```

```python
# Performance comparison
methods = ["Python Loop", "NumPy"]
times = [python_time, numpy_time]

# Create bar chart
plt.figure(figsize=(7, 4))
plt.bar(methods, times)

plt.title("Python Loop vs NumPy Performance")
plt.xlabel("Method")
plt.ylabel("Execution Time (seconds)")

plt.show()
```

```python
# Monthly sales data
monthly_sales = np.array([12000, 15000, 18000, 14000, 20000, 25000])

# Summary report
print("MONTHLY SALES ANALYSIS")
print("----------------------")

print("Total Sales:", np.sum(monthly_sales))
print("Average Sales:", np.mean(monthly_sales))
print("Highest Sales:", np.max(monthly_sales))
print("Lowest Sales:", np.min(monthly_sales))
print("Standard Deviation:", round(np.std(monthly_sales), 2))
print("Median Sales:", np.median(monthly_sales))

# Find months with sales above 15000
print("Sales Above 15000:", monthly_sales[monthly_sales > 15000])
```
